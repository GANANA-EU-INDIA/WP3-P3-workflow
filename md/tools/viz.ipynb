### Energy minimization

Firstly, generate the potential file with:
```bash
printf "Potential\n0\n" | gmx energy -f em.edr -o potential.xvg -xvg none
```
in the protein directory.

````python
path_potential = "/leonardo/home/userinternal/lsalicar/compss_test/gromacs_md/md-tutorial-gromacs/data/output/1fjs/potential.xvg"
table = pd.read_csv(path_potential, sep='\\s+', header=None, names=['step','energy'], skiprows=24)

plt.plot(table["step"], table["energy"])
plt.xlabel("step")
plt.ylabel("energy")
plt.grid(alpha=0.3)

plt.show()
````
